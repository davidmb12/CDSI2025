```python
def time_to_seconds(time_str):
    if isinstance(time_str,str) and ":" in time_str:
        minutes,seconds = map(int,time_str.split(":"))
        return minutes * 60 +seconds
    return np.nan
```

```python
def get_hrv_at_time(file_path,time_index):
    with open(file_path,"r") as file:
        hrv_data = [int(line.strip()) for line in file.readlines()]
    name = (file_path.split("/")[-1]).split(".")[0]
    max_time = last_times.loc[last_times["user"]==name,"time_seconds"].values[0]

    hrv_at_time = (time_index*max_time)//len(hrv_data)
    return hrv_data[hrv_at_time]
    #if 0<=time_index < len(hrv_data):
    #   return hrv_data[time_index]
    return np.nan
```

```python
events ={
    "relax":0,
    "VR on":1,
    "going up in the elevator":2,
    "open elevator doors":3,
    "start walk":4,
    "end walk":5,
    "elevator":6,
    "eleveator":6,
    "start red light":7,
    "end red light":8,
    "fall":9,
    "VR off":10,
}
```

```python
#Load the csv file
file_path = './FISIOLOGICOS/Heart_Rate_VR_Data.csv'
df = pd.read_csv(file_path)
df.head()
```

```python
# Time to seconds
df["time_seconds"] = df["time"].apply(time_to_seconds)
df.head()
```

```python
# Event from string to int
df["event"] = df["event"].apply(get_event)
df.head()
```

```python
# Cargar los datos de ritmo cardiaco para cada usuario
heart_rates_data = {}
for user,filepath in heart_rate_files.items():
    try:
        with open(filepath,"r") as f:
            heart_rates = [int(line.strip()) for line in f if line.strip().isdigit()]
            heart_rates_data[user] = heart_rates
    except Exception as e:
        print(f"Error al leer {filepath}:{e}")
print(heart_rates_data)
```

```python
hrv_feature_list= []
for k,v in heart_rates_data.items():
    features = extract_hrv_features(f"./FISIOLOGICOS/{k}.txt")
   
    hrv_feature_list.append(features)
hrv_features_df = pd.DataFrame(hrv_feature_list)
hrv_features_df.head()
```

```python
df = df.merge(hrv_features_df)
print(df)
```

```python
directory = "./FISIOLOGICOS/"
df["hrv_at_time"] = df.apply(lambda row: get_hrv_at_time(
    os.path.join(directory, row["hear rate file"]), row["time_seconds"]), axis=1)
```

```python
X = df.drop(columns=["hear rate file","fear level"])
y = df["fear level"]
print(df)
# Guardar nombres de características
feature_names = X.columns.tolist()
```

```python
# Dividir datos en conjunto de entrenamiento y prueba
X_train,X_test, y_train,y_test = train_test_split(X,y,test_size = 0.3,random_state = 42)
```

```python
model = RandomForestClassifier(n_estimators = 2000, random_state = 42)
model.fit(X_train,y_train)
```

```python
# Evaluar modelo
y_pred = model.predict(X_test)
print("Accuracy:",accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
def predict_fear_level(file_path, sex=1, vr_experience=1, heart_issue=0, time_seconds=0, hrv_at_time=0,event = 5):
    """Predice el nivel de miedo dado un archivo HRV y características adicionales."""
    new_features = extract_hrv_features(file_path)
    new_features["sex"] = sex
    new_features["VR experience"] = vr_experience
    new_features["heart/vertigo issue"] = heart_issue
    new_features["time_seconds"] = time_seconds
    new_features["hrv_at_time"] = hrv_at_time
    new_features["event"] = event
    
    new_df = pd.DataFrame([new_features])
    
    # Asegurar que las características coincidan con las usadas en el entrenamiento
    new_df = new_df[feature_names]
    
    prediction = model.predict(new_df)
    return prediction[0]
```

```python
# Ejemplo de uso con un nuevo archivo HRV
new_hrv_file = "./FISIOLOGICOS/ignacio.txt"

predicted_fear = predict_fear_level(new_hrv_file)
print(f"Prediccion de nivel de miedo para {new_hrv_file}:{predicted_fear}")
```
